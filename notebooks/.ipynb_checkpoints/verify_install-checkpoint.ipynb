```python
import matplotlib as plt
import seaborn as sb
import numpy as np
import scipy as sp
import sklearn 
import pandas as pd

# Test matplotlib
try:
    plt.pyplot.plot([1, 2, 3], [4, 5, 6])  
    print("matplotlib imported successfully")
except Exception as e:
    print(f"Error importing matplotlib: {e}")
    
# Test seaborn
try:
    sb.set_theme()  
    print("seaborn imported successfully")
except Exception as e:
    print(f"Error importing seaborn: {e}")
    
 # Test numpy
try:
    arr = np.array([1, 2, 3]) 
    print("numpy imported successfully")
except Exception as e:
    print(f"Error importing numpy: {e}")

# Test scipy
try:
    sp_version = sp.__version__  
    print(f"scipy imported successfully, version: {sp_version}")
except Exception as e:
    print(f"Error importing scipy: {e}")
    
 # Test sklearn  
try:
    from sklearn.datasets import load_iris 
    iris = load_iris()
    print("sklearn imported successfully")
except Exception as e:
    print(f"Error importing sklearn: {e}")
    
# Test pandas
try:
    df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})  
    print("pandas imported successfully")
except Exception as e:
    print(f"Error importing pandas: {e}")
```
